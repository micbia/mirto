```python
import numpy as np
import matplotlib.pyplot as plt

from glob import glob

import astropy.units as u
```

```python
arr = ['skalow_AA1_layout.txt', 'skalow_AA2_layout.txt', 'skalow_AAstar_layout.txt', 'skalow_AA4_layout.txt']

fig, axs = plt.subplots(figsize=(10,10), ncols=2, nrows=2, sharex=True, sharey=True)
axs = axs.flatten()

for i, val in enumerate(arr):
    telescope_layout = (np.loadtxt(val) * u.m).to('km')

    axs[i].set_title(val)
    axs[i].scatter(telescope_layout[:,0], telescope_layout[:,1], s=1, marker='.')
    
    axs[i].set_xlabel('x [%s]' %telescope_layout.unit), axs[i].set_ylabel('y [%s]' %telescope_layout.unit)
    axs[i].set_xlim(-50, 50), axs[i].set_ylim(-50, 50)
    
    axs[i].label_outer()
    
plt.savefig('layout_example.png', bbox_inches='tight')
plt.show(), plt.clf()
```
